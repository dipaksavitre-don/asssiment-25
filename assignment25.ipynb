```python
"""Q1. (Manual Hyperparameter Tuning – KNN)
Load the Iris dataset.
Split into train and test sets (test_size=0.33, random_state=42).
Create a KNN model and manually try different values of n_neighbors (e.g., 3, 5, 7, 11, 13, 15).
For each value, train the model and print the test score.
Write which value gave the highest accuracy.
"""
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris

iris = load_iris()

pd.DataFrame(data=iris.data,columns=iris.feature_names)
```

```python


# Convert target integers into species names
species = pd.Categorical.from_codes(iris.target, iris.target_names)

# Create DataFrame with features + species
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = species

df.head()
```

```python
# spliting into trainin and testing
from sklearn.model_selection import train_test_split
x = iris.data
y = iris.target

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.33, random_state=42,shuffle=True)
```

```python
# creating the knn model and trying  to it on different values
from sklearn.neighbors import KNeighborsClassifier

scores = {}
values = [3, 5, 7, 11, 13, 15]

for i in values:
    model = KNeighborsClassifier(n_neighbors=i)
    model.fit(x_train,y_train)
    score = model.score(x_test,y_test)
    scores[i]=score
    print(f"for N_neighbors:{i} accuracy is:{score}")
```

on the iris dataset the values of n_neighbors = 11,13 and 15 gives the highest accuracy it suggest that having the large number of the neighbors classification help our model to perform better

```python
"""Q2. (Manual Hyperparameter Tuning – SVM)
Using the same train-test split from Q1:
Create an SVM model and manually try different combinations of C (1, 10, 20) and kernel (linear, rbf).
Train each combination and print the test accuracy.
Identify the best combination of hyperparameters.
"""

from sklearn.svm import SVC

c = [1, 10, 20]

kernal = ['linear','rbf']

for  i in kernal:
    for j in c:

        model = SVC(C=j,kernel=i)
        model.fit(x_train,y_train)
        score = model.score(x_test,y_test)
        print(f"For kernal :{i} and C :{j} score = {score}")

```

the best combination of the hyperparameters are c = 1 and kernal = linear , c = 10 and kernal = linear and c = 1 kernal = rbf and c= 20 and kernal = rbf

```python
"""Q3. (Grid Search CV)
Apply GridSearchCV on an SVM model:
Create a DataFrame of cv_results_ and display param_C, param_kernel, and
mean_test_score.
Print the best parameters and best score."""

from sklearn.model_selection import GridSearchCV

grid = {'C':[1,10,20],'kernel':['linear','rbf']}

model2 = SVC()

grid_search = GridSearchCV(model2,grid,cv=5)

grid_search.fit(x_train,y_train)

cv_results = pd.DataFrame(grid_search.cv_results_)
best_params = grid_search.best_params_
best_score = grid_search.best_score_

```

```python
"""Q4. (Randomized Search CV)
Apply RandomizedSearchCV on the same SVM model:
Use the same parameter grid as Q3.
Set n_iter=5 and cv=5.
Fit the model.

Display the results DataFrame with important columns.
Compare the best parameters and score with Grid Search results from Q3."""

# applyinh RandomizedSearchCV on the same SVM model

from sklearn.model_selection import RandomizedSearchCV

random_grid = {'C':[1,10,20],'kernel':['linear','rbf']}

model3 = SVC()

random_search = RandomizedSearchCV(model3,random_grid,n_iter=5,cv=5)

random_search.fit(x_train,y_train)

results = pd.DataFrame(random_search.cv_results_)
results[['param_C','param_kernel','mean_test_score']]
```

```python
"""Q6. (Boosting – AdaBoost & Gradient Boosting)
Train the following models:
AdaBoostClassifier (n_estimators=100, random_state=42)
GradientBoostingClassifier (n_estimators=100, learning_rate=0.1,
random_state=42)
Calculate and print the accuracy of both models on the test set."""

from sklearn.ensemble import AdaBoostClassifier, GradientBoostingClassifier

ada_model = AdaBoostClassifier(n_estimators=100,random_state=42)
gb_model = GradientBoostingClassifier(n_estimators=100,learning_rate=0.1,random_state=42)

accuracy_ada = ada_model.fit(x_train,y_train).score(x_test,y_test)
accuracy_gb = gb_model.fit(x_train,y_train).score(x_test,y_test)
print(f"Accuracy of AdaBoost model on test set: {accuracy_ada}")
print(f"Accuracy of Gradient Boosting model on test set: {accuracy_gb}")
```

```python
"""Q7. (Boosting – XGBoost)
Train an XGBClassifier with:
n_estimators=100
learning_rate=0.1
max_depth=3
random_state=42
Calculate and print the test accuracy."""

from xgboost import XGBClassifier

xgb_model = XGBClassifier(n_estimators=100,learning_rate=0.1,max_depth=3,random_state=42)
accuracy_xgb = xgb_model.fit(x_train,y_train).score(x_test,y_test)
print(f"Accuracy of xgboost model on iris dataset is:{accuracy_xgb}")
```

```python
"""Q8. (Hyperparameter Tuning on Random Forest)
Apply GridSearchCV on Random Forest Classifier:
Parameter grid:
n_estimators: [50, 100, 150]
max_depth: [3, 5, 7]
Use cv=5.
Find and print the best parameters and best score."""

from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier

param_grid = {'n_estimators': [50, 100, 150],'max_depth': [3, 5, 7]}
random_forest = RandomForestClassifier(random_state=42)

grid_search_rf = GridSearchCV(random_forest,param_grid,cv=5)
grid_search_rf.fit(x_train,y_train)
print(f"Best Parmeters for random forest:{grid_search_rf.best_params_}")
print(f"Best Score for random forest:{grid_search_rf.best_score_}")
```
